```python
import pandas as pd
data={
    "Name":["A","B","C","D","E"],
    "Marks":[65,72,81,55,90]
}
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

```python
trails=[10,100,1000,10000]

for n in trails:
  result=np.random.choice(["Head","Tail"],size=n)
  probability = np.sum(result=="Head")/n
  print(n,probability)
```

```python
import numpy as np
rolls = np.random.randint(1,7,100)
for number in range(1,7):
  count = np.sum(rolls == number)
  probability = count / 10
  print(number,count,probability)
```
