```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import itertools
```

```python
np.random.seed(42)

# Classical Probability
coin_space = ['H', 'T']
two_coin = list(itertools.product(coin_space, coin_space))
print('Sample Space:', two_coin)
```

```python
two_heads = [s for s in two_coin if s.count('H') == 2]
P_2H = len(two_heads) / len(two_coin)
print(f'P(2 Heads) = {P_2H}')
```

```python
# Dice - Addition Rule
dice = list(range(1, 7))
A = [x for x in dice if x % 2 == 0]   # Even
B = [x for x in dice if x > 3]          # Greater than 3
P_A = len(A)/6;  P_B = len(B)/6
P_AB = len(set(A)&set(B))/6
print(f'P(A∪B) = P(A)+P(B)-P(A∩B) = {P_A+P_B-P_AB:.4f}')
```

```python
#  Empirical Convergence
for n in [10, 100, 1000, 10000]:
    tosses = np.random.binomial(1, 0.5, n)
    print(f'n={n:>6}: P(H)={tosses.mean():.4f}')
```

```python
#  Baye's Theorem
P_Disease = 0.01    # Only 1% of the population has the disease.
P_Pos_D   = 0.99    # Sensitivity - If a person has the disease, the probability that the test is positive is 99%.
P_Pos_ND  = 0.05    # False Positive Rate - If a person does NOT have the disease, there is still a 5% probability that the test is positive.


# Total Probability
P_Pos = P_Pos_D*P_Disease + P_Pos_ND*(1-P_Disease)
# Baye's Theorem
P_D_Pos = (P_Pos_D * P_Disease) / P_Pos
print(f'P(Disease | Positive) = {P_D_Pos:.4f} ({P_D_Pos*100:.2f}%)')
```

```python
# VISUALIZATION
# Empirical convergence plot
n_total = 10000
tosses  = np.random.binomial(1, 0.5, n_total)
emp_prob = [tosses[:i].mean() for i in range(1, n_total+1)]
plt.figure(figsize=(10,4))
plt.plot(emp_prob, color='steelblue', linewidth=1)
plt.axhline(0.5, color='red', linestyle='--', label='True P=0.5')
plt.xscale('log')
plt.title('Empirical Probability Convergence (Law of Large Numbers)')
plt.xlabel('Number of Trials (log scale)')
plt.ylabel('Empirical Probability')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# Exercise Qusetions
""" 1. Coin Toss Probability
Write a Python program to generate the sample space for three coin tosses and calculate the probability of:
- a) Getting exactly 3 Heads
- b) Getting exactly 2 Heads
- c) Getting at least 1 Head
Hint: Use itertools.product()."""


# Generate sample space for 3 coin tosses
sample_space = list(itertools.product(['H', 'T'], repeat=3))
print("Sample Space:")
print(sample_space)

# Total number of outcomes
total = len(sample_space)

# a) Exactly 3 Heads
exactly_3_heads = [outcome for outcome in sample_space
                   if outcome.count('H') == 3]

# b) Exactly 2 Heads
exactly_2_heads = [outcome for outcome in sample_space
                   if outcome.count('H') == 2]

# c) At least 1 Head
at_least_1_head = [outcome for outcome in sample_space
                   if outcome.count('H') >= 1]

# Calculate probabilities
p3 = len(exactly_3_heads) / total
p2 = len(exactly_2_heads) / total
p1 = len(at_least_1_head) / total

print("\nExactly 3 Heads:", exactly_3_heads)
print("Probability of exactly 3 Heads:", p3)

print("\nExactly 2 Heads:", exactly_2_heads)
print("Probability of exactly 2 Heads:", p2)

print("\nAt least 1 Head:", at_least_1_head)
print("Probability of at least 1 Head:", p1)
```

```python
"""2. Dice Probability
Write a Python program to calculate the probability of getting:
- a) An even number
- b) An odd number
- c) A number greater than 4
- d) A number less than or equal to 3
Use the sample space of a single die."""

# Sample space of a single die
sample_space = [1, 2, 3, 4, 5, 6]

# Total number of outcomes
total = len(sample_space)

# a) Even number
even = [x for x in sample_space if x % 2 == 0]

# b) Odd number
odd = [x for x in sample_space if x % 2 != 0]

# c) Number greater than 4
greater_4 = [x for x in sample_space if x > 4]

# d) Number less than or equal to 3
less_equal_3 = [x for x in sample_space if x <= 3]

# Calculate probabilities
p_even = len(even) / total
p_odd = len(odd) / total
p_greater_4 = len(greater_4) / total
p_less_equal_3 = len(less_equal_3) / total

# Display results
print("Sample Space:", sample_space)

print("\nEven numbers:", even)
print("Probability of even number:", p_even)

print("\nOdd numbers:", odd)
print("Probability of odd number:", p_odd)

print("\nNumbers greater than 4:", greater_4)
print("Probability of number greater than 4:", p_greater_4)

print("\nNumbers less than or equal to 3:", less_equal_3)
print("Probability of number less than or equal to 3:", p_less_equal_3)
```

```python
"""3. Two-Dice Experiment
Generate the sample space for rolling two dice and calculate the probability that:
- a) The sum is 7
- b) The sum is greater than 8
- c) Both dice show the same number
- d) At least one die shows 6
Hint: Use itertools.product(range(1,7), range(1,7))."""

# Generate sample space for rolling two dice
sample_space = list(itertools.product(range(1, 7), range(1, 7)))

print("Sample Space:")
print(sample_space)

# Total number of outcomes
total = len(sample_space)

# a) Sum is 7
sum_7 = [outcome for outcome in sample_space
         if outcome[0] + outcome[1] == 7]

# b) Sum is greater than 8
sum_greater_8 = [outcome for outcome in sample_space
                 if outcome[0] + outcome[1] > 8]

# c) Both dice show the same number
same_number = [outcome for outcome in sample_space
               if outcome[0] == outcome[1]]

# d) At least one die shows 6
at_least_one_6 = [outcome for outcome in sample_space
                  if outcome[0] == 6 or outcome[1] == 6]

# Calculate probabilities
p_sum_7 = len(sum_7) / total
p_sum_greater_8 = len(sum_greater_8) / total
p_same = len(same_number) / total
p_at_least_one_6 = len(at_least_one_6) / total

# Display results
print("\nSum is 7:", sum_7)
print("Probability of sum 7:", p_sum_7)

print("\nSum greater than 8:", sum_greater_8)
print("Probability of sum greater than 8:", p_sum_greater_8)

print("\nBoth dice show the same number:", same_number)
print("Probability of same number:", p_same)

print("\nAt least one die shows 6:", at_least_one_6)
print("Probability of at least one 6:", p_at_least_one_6)
```

```python
"""4. Addition Rule with Dice
For a single die, define:
A: Getting an even number
B: Getting a number greater than 3
Write a Python program to calculate:
P(A), P(B),P(A∩B),P(AUB)
Verify the addition rule:
P(A∪B)=P(A)+P(B)−P(A∩B)"""

# Sample space of a single die
sample_space = [1, 2, 3, 4, 5, 6]

# Define events
A = [x for x in sample_space if x % 2 == 0]
B = [x for x in sample_space if x > 3]

# Find intersection A ∩ B
intersection = [x for x in sample_space if x in A and x in B]

# Find union A ∪ B
union = [x for x in sample_space if x in A or x in B]

# Total number of outcomes
total = len(sample_space)

# Calculate probabilities
P_A = len(A) / total
P_B = len(B) / total
P_intersection = len(intersection) / total
P_union = len(union) / total

# Display results
print("Sample Space:", sample_space)
print("Event A:", A)
print("Event B:", B)

print("\nP(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_intersection)
print("P(A ∪ B) =", P_union)

# Verify Addition Rule
calculated = P_A + P_B - P_intersection

print("\nVerification:")
print("P(A ∪ B) =", P_union)
print("P(A) + P(B) - P(A ∩ B) =", calculated)

if P_union == calculated:
    print("Addition Rule Verified")
else:
    print("Addition Rule Not Verified")
```

```python
""" 5. Law of Large Numbers
Simulate 10,000 fair coin tosses and calculate the cumulative probability of Heads after every toss.
Plot:
- X-axis → Number of tosses
- Y-axis → Cumulative probability of Heads
- Horizontal line → \(P(H)=0.5\)
Use a logarithmic X-axis."""

import random
# Number of tosses
n = 10000

# Store cumulative probability of Heads
probability_heads = []

# Count of Heads
heads = 0

# Simulate 10,000 coin tosses
for i in range(1, n + 1):

    # Fair coin toss
    toss = random.choice(['H', 'T'])

    # Count Heads
    if toss == 'H':
        heads = heads + 1

    # Calculate cumulative probability
    probability = heads / i
    probability_heads.append(probability)

# Plot the results
plt.figure(figsize=(10, 6))

# Cumulative probability of Heads
plt.plot(range(1, n + 1), probability_heads,
         label='Cumulative Probability of Heads')

# Horizontal line at P(H) = 0.5
plt.axhline(y=0.5, linestyle='--',
            label='P(H) = 0.5')

# Logarithmic X-axis
plt.xscale('log')

# Labels and title
plt.xlabel('Number of Tosses')
plt.ylabel('Cumulative Probability of Heads')
plt.title('Law of Large Numbers - Coin Toss Experiment')

# Show legend
plt.legend()

# Show grid
plt.grid(True)

# Display plot
plt.show()
```

```python
"""6. Conditional Probability with Dice
Roll a die.
Define:
- \(A\): number is even
- \(B\): number is greater than 3
- Write a program to calculate: P(A|B)
- Verify the result manually."""

# Sample space of a single die
sample_space = [1, 2, 3, 4, 5, 6]

# Define events
A = [x for x in sample_space if x % 2 == 0]
B = [x for x in sample_space if x > 3]

# Find A intersection B
A_intersection_B = [x for x in sample_space
                    if x in A and x in B]

# Calculate probabilities
P_B = len(B) / len(sample_space)
P_A_intersection_B = len(A_intersection_B) / len(sample_space)

# Conditional probability P(A|B)
P_A_given_B = P_A_intersection_B / P_B

# Display results
print("Sample Space:", sample_space)
print("A (Even numbers):", A)
print("B (Numbers greater than 3):", B)
print("A ∩ B:", A_intersection_B)

print("\nP(B) =", P_B)
print("P(A ∩ B) =", P_A_intersection_B)
print("P(A|B) =", P_A_given_B)
```

```python
"""7. Bayes' Theorem - Medical Test Problem
A disease affects 2% of a population.
A diagnostic test has:
- Sensitivity = 95%
- False-positive rate = 4%
- Write a Python program to calculate: P(Disease|Positive)
- Display the answer as a percentage."""

# Given probabilities
P_Disease = 0.02
P_NoDisease = 1 - P_Disease

P_Positive_given_Disease = 0.95
P_Positive_given_NoDisease = 0.04

# Calculate probability of a positive test
P_Positive = (P_Positive_given_Disease * P_Disease) + \
             (P_Positive_given_NoDisease * P_NoDisease)

# Apply Bayes' Theorem
P_Disease_given_Positive = \
    (P_Positive_given_Disease * P_Disease) / P_Positive

# Display result as percentage
print("Probability of Disease given Positive Test:",
      P_Disease_given_Positive * 100, "%")
```

```python
""" 8. Email Classification
- Suppose: 20% of emails are spam.
- A spam email contains the word "offer" with probability 80%.
- A non-spam email contains offer with probability 10%.
- Using Bayes theorem, calculate: P(Spam|offer)
- Write a Python program to display the result."""

# Given probabilities
P_Spam = 0.20
P_NotSpam = 1 - P_Spam

P_Offer_given_Spam = 0.80
P_Offer_given_NotSpam = 0.10

# Calculate probability of getting "offer"
P_Offer = (P_Offer_given_Spam * P_Spam) + \
          (P_Offer_given_NotSpam * P_NotSpam)

# Apply Bayes' Theorem
P_Spam_given_Offer = \
    (P_Offer_given_Spam * P_Spam) / P_Offer

# Display result as percentage
print("P(Spam | offer) =", P_Spam_given_Offer * 100, "%")
```

```python
import math
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# -----------------------------
# Binomial PMF
# -----------------------------
def binomial_probability(n, k, p):
    return math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

# -----------------------------
# Interactive Plotting Function
# -----------------------------
def plot_binomial(n, p):
    # Clear previous plots
    clear_output(wait=True)

    # Re-display the control widgets
    display(ui)

    # Validation
    if n <= 0:
        print("Error: n must be a positive integer.")
        return
    if p < 0 or p > 1:
        print("Error: p must be between 0 and 1.")
        return

    # Calculate values
    x_values = list(range(n + 1))
    probabilities = [binomial_probability(n, x, p) for x in x_values]
    mean = n * p

    # Plotting
    fig, ax = plt.subplots(figsize=(10, 5))
    bars = ax.bar(x_values, probabilities, width=0.7, edgecolor='black', color='skyblue')
    ax.axvline(mean, linestyle='--', color='red', linewidth=2, label=f'Mean = {mean:.2f}')

    # Labels and title
    ax.set_xlabel('Number of successes X', fontweight='bold')
    ax.set_ylabel('Probability P(X)', fontweight='bold')
    ax.set_title(f'Binomial PMF: n = {n}, p = {p}', fontsize=14, fontweight='bold')
    ax.set_xticks(x_values)
    ax.set_ylim(0, max(probabilities) * 1.2)
    ax.legend()
    ax.grid(axis='y', linestyle=':', alpha=0.5)

    plt.tight_layout()
    plt.show()

# -----------------------------
# Create Interactive Widgets
# -----------------------------
n_slider = widgets.IntSlider(value=20, min=1, max=100, step=1, description='n (trials):')
p_slider = widgets.FloatSlider(value=0.35, min=0.0, max=1.0, step=0.01, description='p (prob):')

ui = widgets.HBox([n_slider, p_slider])
out = widgets.interactive_output(plot_binomial, {'n': n_slider, 'p': p_slider})

# Display the interactive component
display(ui, out)
```
